```python
# Codec Technologies Internship Project 1
# Chatbot for Customer Service - Rule Based Chatbot
# Name: [NEE PERU PETTU] | Order ID: 330215

def chatbot_reply(user_input):
    user_input = user_input.lower()

    if "hello" in user_input or "hi" in user_input or "hey" in user_input:
        return "Hello! Welcome to Codec Support. How can I help you today?"
    elif "internship" in user_input or "certificate" in user_input:
        return "Our Internship Certificate will be issued within 24-48 hours after project submission. It contains AICTE ID and QR for verification."
    elif "price" in user_input or "fee" in user_input or "cost" in user_input:
        return "Internship is Free. Only Rs.149 is charged for mentorship and certificate generation."
    elif "duration" in user_input or "time" in user_input:
        return "Internship duration is 1 Month. You can complete it at your own pace."
    elif "project" in user_input:
        return "You need to complete any 2 projects and send 【entity-GitHub¦canonical_name=GitHub】 link to vaishali@codectechnologies.in"
    elif "thank" in user_input:
        return "You're welcome! All the best for your internship."
    elif "bye" in user_input:
        return "Bye! Thank you for chatting with Codec Support."
    else:
        return "Sorry, I didn't understand. Please ask about Internship, Certificate, Fees, or Projects."

print("--- Codec Customer Service Chatbot ---")
print("Type 'bye' to exit")

while True:
    user = input("You: ")
    reply = chatbot_reply(user)
    print("Bot:", reply)
    if "bye" in user.lower():
        break
```

```python
# Codec Technologies Internship Project 2
# Movie Recommendation System - Content Based Filtering
# Name: [NEE PERU PETTU] | Order ID: 330215

movies = {
    "Bahubali": ["Action", "Drama", "Telugu"],
    "RRR": ["Action", "Drama", "Telugu"],
    "Pushpa": ["Action", "Drama", "Telugu"],
    "Kalki": ["Sci-Fi", "Action", "Telugu"],
    "3 Idiots": ["Comedy", "Drama", "Hindi"],
    "Dangal": ["Drama", "Sports", "Hindi"],
    "Interstellar": ["Sci-Fi", "Drama", "English"],
    "Inception": ["Sci-Fi", "Action", "English"],
    "Avengers": ["Action", "Sci-Fi", "English"],
    "KGF": ["Action", "Drama", "Kannada"]
}

def recommend(movie_name):
    movie_name = movie_name.strip().title()
    # Find matching movie - case insensitive
    found = None
    for m in movies:
        if m.lower() == movie_name.lower():
            found = m
            break

    if not found:
        print(f"\nSorry, '{movie_name}' not in our database.")
        print("Available Movies:", ", ".join(movies.keys()))
        return

    print(f"\nYou liked: {found} - Genres: {movies[found]}")
    print("Because you liked this, you may also like:")

    target_genres = set(movies[found])
    scores = []
    for m, genres in movies.items():
        if m == found:
            continue
        common = len(target_genres.intersection(set(genres)))
        if common > 0:
            scores.append((m, common, genres))

    scores.sort(key=lambda x: x[1], reverse=True)
    for m, score, genres in scores[:3]:
        print(f"- {m} | Genres: {genres} | Match Score: {score}")

print("--- Codec Movie Recommendation System ---")
print("Available:", ", ".join(movies.keys()))
user_movie = input("\nEnter a movie you liked: ")
recommend(user_movie)
```
